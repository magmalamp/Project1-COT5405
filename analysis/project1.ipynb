```python


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

```

```python
results = pd.read_csv('C:/Users\shayr\IdeaProjects\Project1-COT5405\greedy-results.csv')

k4 = results[results['k']==4] # both sequences equal in length
k5 = results[results['k']==5] # both sequences equal in length
k6 = results[results['k']==6] # both sequences equal in length

fig, ax = plt.subplots()
ax.scatter(k4["SeqA_Length"].unique(), k4["Runtime_ms"].groupby(k4["SeqA_Length"]).mean(), color='blue')
ax.scatter(k5["SeqA_Length"].unique(), k5["Runtime_ms"].groupby(k5["SeqA_Length"]).mean(), color='green')
ax.scatter(k6["SeqA_Length"].unique(), k6["Runtime_ms"].groupby(k6["SeqA_Length"]).mean(), color='red')
ax.legend(['K = 4','K = 5','K = 6'])
ax.set_xlabel('Residual Length of Sequences')
ax.set_ylabel('Runtime (ms)')
plt.savefig("runtime.png")
plt.show()

## Results for Space ##

fig, ax = plt.subplots()
ax.scatter(k4["SeqA_Length"].unique(), k4["MemUsage_KB"].groupby(k4["SeqA_Length"]).mean(), color='blue')
ax.scatter(k5["SeqA_Length"].unique(), k5["MemUsage_KB"].groupby(k5["SeqA_Length"]).mean(), color='green')
ax.scatter(k6["SeqA_Length"].unique(), k6["MemUsage_KB"].groupby(k6["SeqA_Length"]).mean(), color='red')
ax.legend(['K = 4','K = 5','K = 6'])
ax.set_xlabel('Residual Length of Sequences')
ax.set_ylabel('Memory Usage (KB)')
plt.savefig("memory.png")
plt.show()
```
